## Example From Conjugate Gradient Algorithm in the Documentation for the Project 

```julia
f(x) = (1.0 - x[1])^2 + 100.0 * (x[2] - x[1]^2)^2
function g!(storage, x)
    storage[1] = -2.0 * (1.0 - x[1]) - 400.0 * (x[2] - x[1]^2) * x[1]
    storage[2] = 200.0 * (x[2] - x[1]^2)
end
```

```julia
optimize(f, g!, zeros(2), ConjugateGradient())
# Results of Optimization Algorithm
#  * Algorithm: Conjugate Gradient
#  * Starting Point: [0.0,0.0]
#  * Minimizer: [1.000000002262018,1.0000000045408348]
#  * Minimum: 5.144946e-18
#  * Iterations: 21
#  * Convergence: true
#    * |x - x'| ≤ 0.0e+00: false
#      |x - x'| = 2.09e-10
#    * |f(x) - f(x')| ≤ 0.0e+00 |f(x)|: false
#      |f(x) - f(x')| = 1.55e+00 |f(x)|
#    * |g(x)| ≤ 1.0e-08: true
#      |g(x)| = 3.36e-09
#    * stopped by an increasing objective: false
#    * Reached Maximum Number of Iterations: false
#  * Objective Calls: 54
#  * Gradient Calls: 39
```

## Attempting to modify the Box Minimization example 

```julia
# optimize(d::AbstractObjective, constraints::AbstractConstraints,
#          initial_x::Tx, method::ConstrainedOptimizer, options::Options)
```

```julia
fun(x) = (1.0 - x[1])^2 + 100.0 * (x[2] - x[1]^2)^2

function fun_grad!(g, x)
    g[1] = -2.0 * (1.0 - x[1]) - 400.0 * (x[2] - x[1]^2) * x[1]
    g[2] = 200.0 * (x[2] - x[1]^2)
end

function fun_hess!(h, x)
    h[1, 1] = 2.0 - 400.0 * x[2] + 1200.0 * x[1]^2
    h[1, 2] = -400.0 * x[1]
    h[2, 1] = -400.0 * x[1]
    h[2, 2] = 200.0
end;

x0 = [0.0, 0.0]
df = TwiceDifferentiable(fun, fun_grad!, fun_hess!, x0)

lx = [-0.5, -0.5];
ux = [0.5, 0.5];
dfc = TwiceDifferentiableConstraints(lx, ux)

# Added Optim.Options() to get information about the iterations
res = optimize(df, dfc, x0, IPNewton(), Optim.Options(show_trace = true))
```

```julia
fun(x) = (1.0 - x[1])^2 + 100.0 * (x[2] - x[1]^2)^2

function fun_grad!(g, x)
    g[1] = -2.0 * (1.0 - x[1]) - 400.0 * (x[2] - x[1]^2) * x[1]
    g[2] = 200.0 * (x[2] - x[1]^2)
end

function fun_hess!(h, x)
    h[1, 1] = 2.0 - 400.0 * x[2] + 1200.0 * x[1]^2
    h[1, 2] = -400.0 * x[1]
    h[2, 1] = -400.0 * x[1]
    h[2, 2] = 200.0
end;

x0 = [0.0, 0.0]
df = TwiceDifferentiable(fun, fun_grad!, fun_hess!, x0)

#Going to modify the constraints and see where it converges to
lx = [-1.0, -1.0];
ux = [1.0, 1.0];
dfc = TwiceDifferentiableConstraints(lx, ux)

# Added Optim.Options() to get information about the iterations
res = optimize(df, dfc, x0, IPNewton(), Optim.Options(show_trace = true))
```

## What changed with the changes I made
The final objective value went down, became a smaller value from: 2.500000e-01 to: 6.837831e-17
